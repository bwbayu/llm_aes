```python
import pandas as pd
from transformers import AutoTokenizer
```

```python
df = pd.read_excel("training_set_rel3.xlsx")
# df = df[df['essay_set'] == 1]
df = df[['essay_set', 'essay']]
print(df.info())
df.head()
```

```python
tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")
```

```python
def get_length(text):
    encoding = tokenizer.encode_plus(
            text,
            add_special_tokens=True,
            return_tensors='pt',
        )
    return encoding['input_ids'].flatten().shape[0]

df['essay_len'] = df['essay'].apply(get_length)
```

```python
print("essay 1")
print(df[df['essay_set'] == 1]['essay_len'].describe())

print("essay 2")
print(df[df['essay_set'] == 2]['essay_len'].describe())

print("essay 3")
print(df[df['essay_set'] == 3]['essay_len'].describe())

print("essay 4")
print(df[df['essay_set'] == 4]['essay_len'].describe())

print("essay 5")
print(df[df['essay_set'] == 5]['essay_len'].describe())

print("essay 6")
print(df[df['essay_set'] == 6]['essay_len'].describe())

print("essay 7")
print(df[df['essay_set'] == 7]['essay_len'].describe())

print("essay 8")
print(df[df['essay_set'] == 8]['essay_len'].describe())
```
