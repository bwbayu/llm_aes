```python
import pandas as pd
import numpy as np
import regex as re
```

## Rahumoto Dataset

```python
df = pd.read_csv('../data/aes_dataset_5k_clean.csv')
df = df[df['dataset'] == 'analisis_essay'][['question', 'reference_answer', 'answer', 'score', 'normalized_score', 'dataset', 'dataset_num']]
print(df.info())
df.head()
```

```python
remove_question = ["analisis_essay-1", "analisis_essay-11", "analisis_essay-21", "analisis_essay-31"]
df_seen = df[~df['dataset_num'].isin(remove_question)].reset_index(drop=True)
print(df_seen.info())
df_seen.head()
df_seen.to_csv("aes_dataset_indo.csv", index=False)
```

```python
df_unseen = df[df['dataset_num'].isin(remove_question)].reset_index(drop=True)
print(df_unseen.info())
df_unseen.head()
df_unseen.to_csv("aes_dataset_indo_unseen.csv", index=False)
```

## SAG Dataset

```python
df_sag = pd.read_csv('../data/aes_dataset_5k_clean.csv')
df_sag = df_sag[df_sag['dataset'] == 'sag'][['question', 'reference_answer', 'answer', 'score', 'normalized_score', 'dataset']].reset_index(drop=True)
df_sag["dataset_num"] = df_sag.groupby("dataset")["question"].transform(
    lambda x: ["sag-" + str(i + 1) for i in pd.factorize(x)[0]]
)
print(df_sag.info())
df_sag.head()
```

```python
remove_question = ["sag-75", "sag-76", "sag-77","sag-78", "sag-79", "sag-80", "sag-81", "sag-82"]
df_sag_seen = df_sag[~df_sag['dataset_num'].isin(remove_question)].reset_index(drop=True)
print(df_sag_seen.info())
df_sag_seen.tail()
```

```python
df_sag_unseen = df_sag[df_sag['dataset_num'].isin(remove_question)].reset_index(drop=True)
print(df_sag_unseen.info())
df_sag_unseen.head()
```

## Concat dataset

```python
new_df = pd.concat([df_seen, df_sag_seen], axis=0)
print(new_df.info())
new_df.head()
new_df.to_csv("clean_df.csv", index=False)
```

```python
new_df_unseen = pd.concat([df_unseen, df_sag_unseen], axis=0)
print(new_df_unseen.info())
new_df_unseen.tail()
new_df_unseen.to_csv("clean_df_unseen.csv", index=False)
```
